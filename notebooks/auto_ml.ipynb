```python
from autogluon.tabular import TabularPredictor
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('../data/credit_risk_dataset.csv')
df.sample(5)
```

```python
train_df, test_df = train_test_split(df, random_state= 42, shuffle= True, test_size= 0.3, stratify=df['loan_status'])
print('X_train shape:', train_df.shape)
print('X_test shape:', test_df.shape)
```

```python
predictor = TabularPredictor(label= "loan_status").fit(train_data= train_df, fit_strategy= 'sequential')
```

```python
predictions = predictor.predict(test_df)
predictions
```
